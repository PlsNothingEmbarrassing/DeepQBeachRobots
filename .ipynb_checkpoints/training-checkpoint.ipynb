# Deep Q-learning for beach-cleaning robots

This notebook trains and evaluates a TF-Agents DQN against the Mesa beach environment. The observation is an egocentric local window with separate channels for the robot, rubbish, and obstacles. Run the cells in order; training and policy loading are explicit and do not happen during import.

## 1. Imports and reproducibility

If TensorFlow cannot load its native Windows runtime, repair the selected virtual environment before continuing.

```python
from pathlib import Path
import random

import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
from tf_agents.agents.dqn import dqn_agent
from tf_agents.networks import q_network
from tf_agents.replay_buffers import tf_uniform_replay_buffer
from tf_agents.trajectories import trajectory
from tf_agents.utils import common

from DeepQBeachRobots.environment import MesaGymEnv
from DeepQBeachRobots.robots_Q_learning import collect_step

SEED = 7
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)
print(tf.__version__)
```

## 2. Environment configuration

The notebook uses a smaller curriculum-friendly environment by default. Increase `N_SMALL_RUBBISH` after the agent learns the basic pickup behavior.

```python
N_OBSTACLES = 8
N_SMALL_RUBBISH = 40
GRID_WIDTH = 20
GRID_HEIGHT = 20
OBSERVATION_RADIUS = 5
MAX_STEPS = 500

train_env = MesaGymEnv(
    n_obstacles=N_OBSTACLES,
    n_small_rubbish=N_SMALL_RUBBISH,
    width=GRID_WIDTH,
    height=GRID_HEIGHT,
    seed=SEED,
    max_steps=MAX_STEPS,
    observation_radius=OBSERVATION_RADIUS,
)
time_step = train_env.reset()
print("action spec:", train_env.action_spec)
print("observation spec:", train_env.observation_spec)
print("observation shape:", time_step.observation.shape)
```

```python
plt.figure(figsize=(4, 4))
plt.imshow(time_step.observation.numpy())
plt.title("Local observation window")
plt.axis("off");
```

## 3. Build the DQN and replay buffer

```python
q_net = q_network.QNetwork(
    train_env.observation_spec,
    train_env.action_spec,
    fc_layer_params=(256, 128),
)
optimizer = tf.keras.optimizers.Adam(learning_rate=1e-4)
train_step_counter = tf.Variable(0)
agent = dqn_agent.DqnAgent(
    train_env.time_step_spec,
    train_env.action_spec,
    q_network=q_net,
    optimizer=optimizer,
    td_errors_loss_fn=common.element_wise_squared_loss,
    train_step_counter=train_step_counter,
)
agent.initialize()

replay_buffer = tf_uniform_replay_buffer.TFUniformReplayBuffer(
    data_spec=agent.collect_data_spec,
    batch_size=1,
    max_length=50_000,
)
dataset = replay_buffer.as_dataset(
    num_parallel_calls=3,
    sample_batch_size=32,
    num_steps=2,
).prefetch(3)
iterator = iter(dataset)
```

## 4. Warm up the replay buffer

A warm-up phase prevents the first updates from being based on a nearly empty buffer.

```python
INITIAL_COLLECT_STEPS = 1_000
for _ in range(INITIAL_COLLECT_STEPS):
    collect_step(train_env, agent, replay_buffer)
print("replay buffer size:", replay_buffer.num_frames().numpy())
```

## 5. Train

The history records moving-average loss and episode-level metrics so training quality can be inspected rather than inferred from a single loss value.

```python
NUM_ITERATIONS = 10_000
LOG_INTERVAL = 250
losses = []
episode_metrics = []

for iteration in range(NUM_ITERATIONS):
    collect_step(train_env, agent, replay_buffer)
    experience, _ = next(iterator)
    loss = float(agent.train(experience).loss)
    losses.append(loss)

    if train_env.current_time_step().is_last():
        episode_metrics.append(train_env.episode_metrics.copy())

    if iteration % LOG_INTERVAL == 0:
        recent = losses[-100:]
        print(
            f"step={iteration:05d} "
            f"loss={np.mean(recent):.4f} "
            f"episodes={len(episode_metrics)}"
        )
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(losses, alpha=0.25)
if len(losses) >= 100:
    moving_loss = np.convolve(losses, np.ones(100) / 100, mode="valid")
    axes[0].plot(np.arange(99, len(losses)), moving_loss)
axes[0].set_title("Training loss")
axes[0].set_xlabel("Training iteration")
axes[0].set_ylabel("Loss")

if episode_metrics:
    axes[1].plot([m["picked"] for m in episode_metrics], label="picked")
    axes[1].plot([m["failed_moves"] for m in episode_metrics], label="failed moves")
    axes[1].set_title("Episode metrics")
    axes[1].legend()
plt.show()
```

## 6. Evaluate and save the policy

```python
evaluation_env = MesaGymEnv(
    n_obstacles=N_OBSTACLES,
    n_small_rubbish=N_SMALL_RUBBISH,
    width=GRID_WIDTH,
    height=GRID_HEIGHT,
    seed=SEED + 10_000,
    max_steps=MAX_STEPS,
    observation_radius=OBSERVATION_RADIUS,
)
evaluation_metrics = []
for _ in range(10):
    time_step = evaluation_env.reset()
    while not time_step.is_last():
        action_step = agent.policy.action(time_step)
        time_step = evaluation_env.step(action_step.action)
    evaluation_metrics.append(evaluation_env.episode_metrics.copy())

print({
    key: float(np.mean([metrics[key] for metrics in evaluation_metrics]))
    for key in evaluation_metrics[0]
})

policy_path = Path("policies") / "beach_dqn"
policy_path.parent.mkdir(exist_ok=True)
tf.saved_model.save(agent.policy, str(policy_path))
print("saved:", policy_path)
```
